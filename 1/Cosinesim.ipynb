```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

```python
sentences = [
    "I love machine learning",
    "AI is transforming the world",
    "Dogs are great pets",
    "Cats are independent animals",
    "Python is a programming language",
    "Deep learning uses neural networks",
    "I enjoy going to the gym",
    "Cloud computing is scalable",
    "Transformers are powerful models",
    "FastAPI is used for building APIs",
    "I enjoy learning how artificial intelligence work"
]

model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = model.encode(sentences)
```

```python
print(embeddings.shape) 
```

```python
def cosine_similarity(a, b):
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
```

```python
for i in range(1, len(sentences)):
    score = cosine_similarity(embeddings[0], embeddings[i])
    print(f"Similarity with '{sentences[i]}': {score:.4f}")
```
